```python

import pandas as pd

df = pd.read_csv("DataAnalytics-L1-EDARetailSales/retail_sales_dataset.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python

# Check missing values
print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python

# Select numerical columns
numeric_columns = ["Age", "Quantity", "Price per Unit", "Total Amount"]

# Calculate statistics
print("Mean:")
print(df[numeric_columns].mean())

print("\nMedian:")
print(df[numeric_columns].median())

print("\nMode:")
print(df[numeric_columns].mode().iloc[0])

print("\nStandard Deviation:")
print(df[numeric_columns].std())
```

```python

# Convert Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

print("Date conversion completed!")
print("Date data type:", df["Date"].dtype)
print("Invalid dates:", df["Date"].isna().sum())
```

```python

# Group sales by month
df["Month"] = df["Date"].dt.to_period("M")

monthly_sales = df.groupby("Month")["Total Amount"].sum()

print("Monthly Sales:")
print(monthly_sales)

# Create line chart
plt.figure(figsize=(12, 5))
plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

# Count transactions by gender
gender_counts = df["Gender"].value_counts()

print("Transactions by Gender:")
print(gender_counts)

# Create bar chart
plt.figure(figsize=(7, 5))
gender_counts.plot(kind="bar")

plt.title("Transactions by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Display customer age statistics
print("Customer Age Statistics:")
print(df["Age"].describe())

# Create histogram
plt.figure(figsize=(8, 5))
plt.hist(df["Age"], bins=10, edgecolor="black")

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

```python

# Count transactions by product category
category_counts = df["Product Category"].value_counts()

print("Transactions by Product Category:")
print(category_counts)

# Create bar chart
plt.figure(figsize=(9, 5))
category_counts.plot(kind="bar")

plt.title("Transactions by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Calculate total revenue by product category
category_sales = (
    df.groupby("Product Category")["Total Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Revenue by Product Category:")
print(category_sales)

# Create bar chart
plt.figure(figsize=(9, 5))
category_sales.plot(kind="bar", color="teal")

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Select numerical columns
numeric_data = df[
    ["Age", "Quantity", "Price per Unit", "Total Amount"]
]

# Calculate correlation
correlation_matrix = numeric_data.corr()

print("Correlation Matrix:")
print(correlation_matrix)

# Create heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python

# Step 16: Quarterly Sales Trend

# Group sales by quarter
df["Quarter"] = df["Date"].dt.to_period("Q")

quarterly_sales = df.groupby("Quarter")["Total Amount"].sum()

# Display quarterly sales
print("Quarterly Sales:")
print(quarterly_sales)

# Create line chart
plt.figure(figsize=(9, 5))
plt.plot(
    quarterly_sales.index.astype(str),
    quarterly_sales.values,
    marker="o"
)

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

# Step 17: Total Sales by Gender

gender_sales = df.groupby("Gender")["Total Amount"].sum()

print("Total Sales by Gender:")
print(gender_sales)

# Create bar chart
plt.figure(figsize=(7, 5))
gender_sales.plot(kind="bar", color="skyblue")

plt.title("Total Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
